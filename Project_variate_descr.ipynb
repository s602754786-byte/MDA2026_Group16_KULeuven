## Dataset Overview

This project uses three complementary tables:

- `Counts`: the main fact table, containing the observed counts over time
- `Directions`: a lookup table explaining the meaning of `IN`, `OUT`, and `IN/OUT` at each site
- `Sites`: a metadata table describing the counting locations

### 1. Counts table

This is the core time-series table. Each row represents one counting result for a given site, direction, object type, and time interval.

- `site ID`: identifier of the counting site. This is a foreign key linking the record to the corresponding site in the `Sites` table.
- `richting`: traffic direction. Possible values are `IN`, `OUT`, or `IN/OUT`.
  - `IN`: traffic moving in one direction
  - `OUT`: traffic moving in the opposite direction
  - `IN/OUT`: combined or non-directional count
- `type`: type of object detected by the sensor.  
  Common values include:
  - `FIETSERS`: cyclists
  - `VOETGANGERS`: pedestrians
  - `AUTOS`: cars
  - `BUSSEN`: buses
  - `MOTORFIETSEN`: motorcycles
  - and several other categories
- `van`: start time of the counting interval
- `tot`: end time of the counting interval
- `aantal`: number of detected objects during that interval

Note: If the goal is to analyze bicycle traffic, the data should usually be filtered to `type = FIETSERS`.

### 2. Directions table

This table helps interpret the direction variable used in the `Counts` table.

- `site ID`: identifier of the counting site
- `richting`: direction code (`IN`, `OUT`, or `IN/OUT`)
- `naam`: textual description of the direction, for example "towards the city center"

This table is useful because `IN` and `OUT` are abstract labels. Their real-world meaning may differ across sites.

### 3. Sites table

This table contains metadata about the counting stations themselves.

- `site ID`: primary key of the counting site
- `site nr`: EcoCounter device identifier
- `long`: longitude of the site in WGS84 coordinates
- `lat`: latitude of the site in WGS84 coordinates
- `naam`: name of the site
- `domein`: name of the domain or area where the site is located
- `wegnr`: road number
- `district`: road district number
- `gemeente`: municipality of the site
- `interval`: counting interval in minutes
- `datum_van`: date when the sensor was installed or became active

### Table Relationships

The tables can be linked as follows:

- `Counts.site ID` -> `Sites.site ID`
- `Counts.(site ID, richting)` -> `Directions.(site ID, richting)`

Together, these tables allow us to analyze bicycle traffic from temporal, directional, and spatial perspectives.
